# Operaciones entre columnas de un DataFrame

```python
# Supongamos que el precio de venta es con impuestos... cómo obtengo los precios
# de las laptops antes de impuestos.. suponiendo un impuesto de 16%

datos["Price"]/1.16
```

```python
# Cómo creamos una nueva serie en el DataFrame
datos["Price before taxes"]=datos["Price"]/1.16
```

```python
datos.head()
```

```python
# Cuánto paga la tienda por impuestos en la venta de computadoras
datos["Taxes"]=datos["Price"]-datos["Price before taxes"]
datos.head()
```

```python
datos["Taxes"].sum()  # Cantidad total a pagar por impuestos
```

```python
datos["Taxes"].mean() # Impuestos promedio
```

# Ordenar un DataFrame
Sintaxis:

nombre_dataframe.sort_values(by="nombre_serie")


IMPORTANTE: Estos cambios no afectan el dataframe original, solo aplican para desplegar el
resultado en la pantalla. Si se desea que los cambios se apliquen al DataFrame original
se debe emplear el parámetro _inplace_=True

```python
datos.sort_values(by="brand")  # Ordena por marca en orden creciente
```

```python
datos.sort_values(by="Price")  # Ordena por precio en sentido creciente 
```

```python
datos.sort_values(by="Price", ascending=False)  # Ordena por precio en sentido decreciente 
```

```python
# Para ordenar el DataFrame utilizando varios criterios
# Ordenar por marca y luego por precio

datos.sort_values(by=["brand", "rating","Price"])
```

```python
datos  # Observamos que la ordenación del dataframe sigue siendo la misma, es decir
       # nos mostró una copia del dataframe ordenado.
```

```python
# Para que el orden se aplique al dataframe original
datos.sort_values(by=["brand", "rating","Price"], inplace=True)
```

```python
datos  #ahora si ya el DataFrame está ordenado
```
