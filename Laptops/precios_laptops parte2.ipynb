### Recodificación de una variable, crear una variable a partir de otra

Definimos una computadora "barata" si tiene un Precio es menor a los 50,000
y decimos que es "cara" en otro caso. 


Utilizamos el método where de Numpy...
sintaxis:

np.where(condición, valor en caso de condición==True,valor en caso de condición==False)

IMPORTANTE: Solo recodificación binaria

```python
np.where(datos["Precio"]<=50000,"barata","cara")
# Cómo agraegamos estos resultados al DataFrame
datos["Tipo_laptop"]=np.where(datos["Precio"]<=50000,"barata","cara")
```

```python
datos.head()
```

```python
# Recodificación múltiple...
datos["Precio"].describe()
```

Para una clasificación múltiple utilizamos el método select de Numpy.
condiciones=[condicion1, condicion2, ..., condicionesN]
valores=[valor1,valor2,....,valorN]

```python
# NOTA: En pandas la "y" la intersección se denota por &
#                 la "o" la unión se denota por |
condiciones=[datos["Precio"]<=46095,
             (datos["Precio"]>46095)&(datos["Precio"]<=64990),
             (datos["Precio"]>64990)&(datos["Precio"]<=89636),
             datos["Precio"]>=89636]
valores=["Muy barata","Barata", "Cara", "Muy Cara"]
datos["Tipo_laptop"]=np.select(condiciones,valores)
```
