# Cómo tomar una muestra de un DataFrame

```python
# Empleamos el método .sample() de pandas.
# Extraer una muestra de 10 computadoras, elegidas al azar del DataFrame

datos.sample(10)  # Devuelve 10 laptos elegidas al azar
```

```python
# Para tener reproducibilidad en los resultados, es decir que siempre se elijan las mismas laptos,
# agregamos el parametro random_state (valor semilla para la generación de los números aleatorios)
datos.sample(10, random_state=999)
```

```python
# Cuando deseamos elegir por porcentaje y no por cantidad, utilizamos el parámetro frac
datos.sample(frac=0.20,random_state=999)
```

```python
# Para obtener una muestra de 2000 laptos del dataframe
datos.sample(2000, replace=True)
```

```python
datos["ram_gb"].sample(10)  # Esta es una muestra de la capacidad de memoria en ram de 10 computadoras 
```

```python
datos["brand"].sample(10)  # Esta es una muestra de las marcas de 10 computadoras 
```

```python
# Una copia del la muestra y la guardamos en un nuevo dataFrame 
datos_muestra=datos.sample(frac=0.20, replace=True)  # Muestreo aleatorio con reemplazo del 20% del dataframe 
datos_muestra
```

```python
datos_muestra=datos.sample(frac=0.20, ignore_index=True)  # Muestreo aleatorio sin reemplazo del 20% del dataframe 
datos_muestra                                             # ignore_index=True reinicia la numeración en el nuevo o en la copia del
                                                          # DataFRame   
```

# Filtrado de un DataFrame: Utilizando el método query()


```python
# Seleccionar la computadoras de la marca DELL...
datos.query("brand=='DELL'")
```

```python
# Seleccionar las computadoras que cuestan más de 100000
datos.query("Price>100000")
```

```python
# Seleccionar las computadoras que cuestan más de 50000 y menos de 70000
datos.query("Price>50000 and Price<=70000")
```

```python
# Seleccionar las computadoras que cuestan más de 50000 y menos de 70000, pero de la marca Lenovo
datos.query("(Price>50000 and Price<=70000) and (brand=='Lenovo')")
```

```python
# Como serían las instrucciones anteriores con un filtrado como lo vimos antes.. 
datos[datos["brand"]=="Lenovo"]
#datos.query("brand=='Lenovo'")
```

```python
# que el precio esté entre 50,000 y menos de 70,000
#datos[(datos["Price"]>50000) & (datos["Price"]<70000)]
datos.query("Price>50000 and Price<70000")
```

```python
# Vamos a definir una función que reciba de entrada un valor (string) y lo devuelva en mayúsculas

def Mayúsculas(c):
    return c.upper()
```
