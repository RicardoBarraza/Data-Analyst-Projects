# Proyecto: Precios de Laptos (gráficos)

```python
# Módulos PANDAS y NUMPY
import pandas as pd
import numpy as np

# Módulos de gráficas
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Jalar el data frame
datos=pd.read_csv("laptopPrice.csv")
# Ojo con las diagonales cuando tengan que definir otra ruta..
# pd.read_csv("C:/Users/migue/Documents/Diplomado_ciencia_datos/datos/nombre_archivo.csv")
```

```python
datos.head()
```

```python
datos.describe()  # Estadísticas de resumen de las variables númericas
```

```python
# Diagrama de cajas y bigotes.
plt.title("Diagrama de caja y bigotes del Precio")
plt.boxplot(datos["Price"])
plt.show()
```

```python
plt.title("Distribución de los precios de laptops")
plt.hist(datos["Price"],bins=35)
plt.show()
```

```python
(datos["brand"].value_counts(normalize=True)*100).round(1)
```

```python
#Para hacer la gráfica de pastel...
ventas=(datos["brand"].value_counts(normalize=True)*100).round(1)
plt.title("Porcentajes de ventas por marca de Laptop")
plt.pie(x=ventas,labels=ventas.index)
plt.show()
```

```python
# Para hacer una grafica de barras
plt.bar(x=ventas.index,height=ventas)
plt.show()
```

# Histogramas a detalle.

Vamos a suponer que lanzamos dos datos y nos fijamos en la suma de los puntos que obtenemos. 
¿Cuales serán los resultados o el resultado más creíble?

```python
# Utilizar el modulo random de numpy
dado1=np.random.randint(1,7,size=10000)
dado2=np.random.randint(1,7,size=10000)
puntaje=dado1+dado2
puntaje
```

```python
plt.title("Histograma de Frecuencias")
plt.hist(puntaje,bins=11)
plt.show()
```

```python
plt.title("Histograma de Frecuencias Relativa")
plt.hist(puntaje,bins=11, density=True)
plt.show()
```
